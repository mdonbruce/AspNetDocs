# Mini‑Lab 1 — Instructor EXECUTED

```
import pandas as pd, numpy as np
DATA_PATH=r'/content/retail_transactions_large.csv'
df=pd.read_csv(DATA_PATH)
print('raw', df.shape)
df.head()
```

```

df2=df.copy()
df2['category']=df2['category'].astype('string')
df2['region']=df2['region'].astype('string').fillna('').str.strip()
for c in ['quantity','unit_price','discount']:
    df2[c]=pd.to_numeric(df2[c], errors='coerce')
df2['discount']=df2['discount'].fillna(0.0).clip(0,0.60)
df2=df2[df2['category'].notna() & (df2['region']!='')]
df2=df2[df2['quantity'].notna() & (df2['quantity']>0)]
df2=df2[df2['unit_price'].notna() & (df2['unit_price']>=0)]
df2['line_total']=(df2['quantity']*df2['unit_price']*(1-df2['discount'])).round(2)
out=df2.groupby('category',as_index=False)['line_total'].sum().sort_values('line_total',ascending=False)
out.head(10)
```
