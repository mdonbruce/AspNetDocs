# Mini‑Lab 2 — Instructor EXECUTED

```
import pandas as pd
DATA_PATH=r'retail_transactions_large.csv'
df=pd.read_csv(DATA_PATH)
print(df.shape)
df.head()
```

```

df2=df.copy()
for c in ['quantity','unit_price','discount']:
    df2[c]=pd.to_numeric(df2[c], errors='coerce')
df2['discount']=df2['discount'].fillna(0.0).clip(0,0.60)
df2['region']=df2['region'].astype('string').fillna('').str.strip()
df2=df2[df2['region']!='']
df2=df2[df2['quantity'].notna() & (df2['quantity']>0)]
df2=df2[df2['unit_price'].notna() & (df2['unit_price']>=0)]
df2['line_total']=(df2['quantity']*df2['unit_price']*(1-df2['discount'])).round(2)
top=df2.groupby('region',as_index=False)['line_total'].sum().sort_values('line_total',ascending=False)
top
```
