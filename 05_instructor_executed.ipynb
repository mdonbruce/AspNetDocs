# Lab 5: Secure Error Logging and Reporting — Instructor (Executed)

**Objective:** Log technical errors safely while showing user-friendly messages.

**Dataset:** `system_events_large.csv`

This notebook is a complete reference solution with outputs.

```python
import pandas as pd

df = pd.read_csv('system_events_large.csv')
df.head()
```

```python
import logging, traceback
logging.basicConfig(filename='app.log', level=logging.INFO, format='%(asctime)s %(levelname)s %(message)s')
logging.info('Logging initialized')
```

```python
def safe_user_message(row):
    lvl = row.get('level','INFO')
    if lvl in ['ERROR','CRITICAL']:
        return 'Something went wrong while processing your request. Please try again.'
    return 'Event processed successfully.'
```

```python
def process_event(row):
    if row['event_type'] == 'db_event' and row['level'] in ['ERROR','CRITICAL']:
        return 1/0
    return True

failures = 0
for _, row in df.head(200).iterrows():
    try:
        process_event(row)
        logging.info(f"processed event_id={row['event_id']} service={row['service']} level={row['level']}")
        print(safe_user_message(row))
    except Exception as e:
        failures += 1
        logging.error(f"event_id={row['event_id']} failed: {type(e).__name__}: {e}")
        logging.debug(traceback.format_exc())
        print(safe_user_message(row))
print('Failures in sample:', failures)
```
